```python
import cv2 #computer vision
import numpy as np
import matplotlib.pyplot as plt

# advanced image processing for automotive vision

#1. read the input road image
image = cv2.imread("colab.jpg")
if image is None:
  raise FileNotFoundError("colab.jpg not found.Check the image path")

#convert BGR to RGB FOR DISPLAYING with matplotlib
rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(rgb)
plt.axis("off")
plt.show()
```

```python
# 2. image enhancement using CLAHE

#convert image to LAB color space
lab = cv2.cvtColor(image, cv2.COLOR_BGR2LAB)
#lab = cv2.cvtColor(rgb,cv2.COLOR_BGR2LAB)
l, a, b = cv2.split(lab)

#contrast limited adaptive Histogram Equilization
Clahe = cv2.createCLAHE(
    clipLimit=2.0,
    tileGridSize=(8,8)
)
```

```python
#cintrast limited adaptive histogram equilization
clahe = cv2.createCLAHE(
    clipLimit=2.0,
    tileGridSize=(8,8)
)
l_enhanced = clahe.apply(l)

#merge channels
enhanced_lab = cv2.merge((l_enhanced, a,b))

# convert back to BGR
enhanced = cv2.cvtColor(enhanced_lab, cv2.COLOR_LAB2BGR)
plt.imshow(enhanced)
plt.axis("off")
plt.show()
```
